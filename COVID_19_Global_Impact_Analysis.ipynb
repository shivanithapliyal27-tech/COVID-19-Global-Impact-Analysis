```python
import pandas as pd
import numpy as np

daily_df = pd.read_csv("worldometer_coronavirus_daily_data.csv")

summary_df = pd.read_csv("worldometer_coronavirus_summary_data.csv")

print("Daily Dataset:", daily_df.shape)
print("Summary Dataset:", summary_df.shape)

display(daily_df.head())
display(summary_df.head())
```

```python
print("DAILY DATASET")
print("Shape:", daily_df.shape)
print("\nColumns:")
print(daily_df.columns.tolist())
print("\nData Types:")
print(daily_df.dtypes)

print("\nSUMMARY DATASET")
print("Shape:", summary_df.shape)
print("\nColumns:")
print(summary_df.columns.tolist())
print("\nData Types:")
print(summary_df.dtypes)
```

```python
print("Daily Dataset Missing Values:")
print(daily_df.isnull().sum())

print("\nSummary Dataset Missing Values:")
print(summary_df.isnull().sum())
```

```python
print("DAILY DATASET STATISTICS")
display(daily_df.describe())

print("\nSUMMARY DATASET STATISTICS")
display(summary_df.describe())
```

```python
print("Total Countries in Daily Dataset:",
      daily_df['country'].nunique())

print("Total Countries in Summary Dataset:",
      summary_df['country'].nunique())

daily_df['date'] = pd.to_datetime(daily_df['date'])

print("\nDate Range:")
print("Start Date:", daily_df['date'].min())
print("End Date:", daily_df['date'].max())

print("\nTop 10 Countries by Total Cases:")
display(
    summary_df[['country', 'total_confirmed']]
    .sort_values(by='total_confirmed', ascending=False)
    .head(10)
)
```

```python
daily_clean = daily_df.copy()
summary_clean = summary_df.copy()

print("Dataset copies created successfully!")
```

```python
print("Daily duplicates:",
      daily_clean.duplicated().sum())

print("Summary duplicates:",
      summary_clean.duplicated().sum())

daily_clean = daily_clean.drop_duplicates()
summary_clean = summary_clean.drop_duplicates()

print("\nDuplicates removed successfully!")
```

```python
print("Daily Missing Values:")
display(daily_clean.isnull().sum())

print("\nSummary Missing Values:")
display(summary_clean.isnull().sum())
```

```python
daily_clean['date'] = pd.to_datetime(
    daily_clean['date'], errors='coerce'
)

print(daily_clean['date'].dtype)
```

```python
daily_clean['country'] = (
    daily_clean['country'].astype(str).str.strip()
)

summary_clean['country'] = (
    summary_clean['country'].astype(str).str.strip()
)

print("Country names standardized!")
```

```python
print("Daily Cleaned Shape:", daily_clean.shape)
print("Summary Cleaned Shape:", summary_clean.shape)

print("\nDaily Dataset:")
display(daily_clean.head())

print("\nSummary Dataset:")
display(summary_clean.head())
```

```python
print("Daily Cleaned Shape:", daily_clean.shape)
print("Summary Cleaned Shape:", summary_clean.shape)

print("\nDaily Dataset:")
display(daily_clean.head())

print("\nSummary Dataset:")
display(summary_clean.head())
```

```python
print("Daily Dataset Columns:")
print(daily_clean.columns.tolist())

print("\nSummary Dataset Columns:")
print(summary_clean.columns.tolist())
```

```python
total_cases = summary_clean['total_confirmed'].sum()

total_deaths = summary_clean['total_deaths'].sum()

total_recovered = summary_clean['total_recovered'].sum()

total_active = summary_clean['active_cases'].sum()

death_rate = (total_deaths / total_cases) * 100

recovery_rate = (total_recovered / total_cases) * 100

print("GLOBAL COVID-19 KPIs")
print("------------------------")
print("Total Confirmed Cases:", f"{total_cases:,.0f}")
print("Total Deaths:", f"{total_deaths:,.0f}")
print("Total Recovered:", f"{total_recovered:,.0f}")
print("Total Active Cases:", f"{total_active:,.0f}")
print("Death Rate:", round(death_rate, 2), "%")
print("Recovery Rate:", round(recovery_rate, 2), "%")
```

```python
top_10_cases = summary_clean.sort_values(
    by='total_confirmed',
    ascending=False
).head(10)

display(top_10_cases[
    ['country', 'total_confirmed', 'total_deaths']
])
```

```python
continent_analysis = summary_clean.groupby(
    'continent'
).agg(
    Total_Cases=('total_confirmed', 'sum'),
    Total_Deaths=('total_deaths', 'sum'),
    Total_Recovered=('total_recovered', 'sum')
).reset_index()

continent_analysis = continent_analysis.sort_values(
    by='Total_Cases', ascending=False
)

display(continent_analysis)
```

```python
daily_clean['date'] = pd.to_datetime(daily_clean['date'])

daily_clean['Year_Month'] = (
    daily_clean['date'].dt.to_period('M').astype(str)
)

monthly_analysis = daily_clean.groupby(
    'Year_Month'
).agg(
    Daily_New_Cases=('daily_new_cases', 'sum'),
    Daily_New_Deaths=('daily_new_deaths', 'sum')
).reset_index()

display(monthly_analysis.head(15))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))

plt.plot(
    monthly_analysis['Year_Month'],
    monthly_analysis['Daily_New_Cases']
)

plt.title('Monthly COVID-19 New Cases Trend')
plt.xlabel('Month')
plt.ylabel('New Cases')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_10_cases.to_csv('COVID_Top_10_Countries.csv', index=False)

continent_analysis.to_csv(
    'COVID_Continent_Analysis.csv', index=False
)

monthly_analysis.to_csv(
    'COVID_Monthly_Analysis.csv', index=False
)

print("All analysis files saved successfully!")
```

```python
country_analysis = summary_clean.groupby(
    'country'
).agg(
    Total_Cases=('total_confirmed', 'sum'),
    Total_Deaths=('total_deaths', 'sum'),
    Total_Recovered=('total_recovered', 'sum'),
    Active_Cases=('active_cases', 'sum')
).reset_index()

country_analysis['Death_Rate'] = (
    country_analysis['Total_Deaths'] /
    country_analysis['Total_Cases'] * 100
)

country_analysis = country_analysis.sort_values(
    by='Total_Cases', ascending=False
)

display(country_analysis.head(10))
```

```python
continent_analysis = summary_clean.groupby(
    'continent'
).agg(
    Total_Cases=('total_confirmed', 'sum'),
    Total_Deaths=('total_deaths', 'sum'),
    Total_Recovered=('total_recovered', 'sum'),
    Active_Cases=('active_cases', 'sum')
).reset_index()

display(continent_analysis)
```

```python
monthly_analysis = daily_clean.groupby(
    'Year_Month'
).agg(
    New_Cases=('daily_new_cases', 'sum'),
    New_Deaths=('daily_new_deaths', 'sum')
).reset_index()

display(monthly_analysis.head())
```

```python
country_analysis.to_csv(
    'COVID_Country_Analysis.csv', index=False
)

continent_analysis.to_csv(
    'COVID_Continent_Analysis.csv', index=False
)

monthly_analysis.to_csv(
    'COVID_Monthly_Trend.csv', index=False
)

daily_clean.to_csv(
    'COVID_Daily_Cleaned.csv', index=False
)

summary_clean.to_csv(
    'COVID_Summary_Cleaned.csv', index=False
)

print("All Power BI datasets exported successfully!")
```
